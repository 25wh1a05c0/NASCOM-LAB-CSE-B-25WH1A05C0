**FUNCTIONS**

```python
#BASIC FUNCTION
def greet():
    print("Hello! Welcome to Python Functions.")
greet()
```

```python
#FUNCTION WITH PARAMETERS
def greet_user(name):
    print(f"Hello, {name}!")
greet_user("Alice")
greet_user("Bob")
```

```python
#FUNCTION WITH DEFAULT PARAMETERS
def greet_user(name="Guest"):
    print(f"Welcome, {name}!")
greet_user()
greet_user("Ravi")
```

```python
#USING *args
def print_names(*names):
    print("Names passed:", names)
    for name in names:
        print("Hello", name)
print_names("Alice", "bob", "Charlie")
```

```python
#USING **kwargs
def print_info(**info):
    print("User Info:")
    for key, value in info.items():
        print(f"{key} = {value}")
print_info(name="Alice", age=25, city="Delhi")
```

```python
#COMBINING *args AND **kwargs
def show_details(*args, **kwargs):
    print("Args:", args)
    print("Kwargs:", kwargs)
show_details(1, 2, 3, name="Ravi", city="Mumbai")
```

```python
#FUNCTION NESTING
def multiply(a, b):
    return a * b
def square(n):
    return multiply(n, n)
print("Square of 4:", square(4))
```

```python
#LISTS
fruits = ['apple', 'banana', 'cherry', 'apple']
print("Original list:", fruits)
print("First fruit:", fruits[0])
```

```python
# Adding an item
fruits.append('orange')
print("After appending:", fruits)
```

```python
# Removing an item
fruits.remove('banana')
print("After removing banana:", fruits)
# List length
print("Number of fruits:", len(fruits))
```

```python
#TUPLES
dimensions = (10, 20, 30)
print("Tuple:", dimensions)
```

```python
# Accessing elements
print("Second element:", dimensions[1])
```

```python
#SETS
unique_numbers = {1, 2, 3, 2, 1}
print("Set (no duplicates):", unique_numbers)
```

```python
# Adding elements
unique_numbers.add(4)
print("After adding 4:", unique_numbers)
```

```python
# Removing elements
unique_numbers.remove(2)
print("After removing 2:", unique_numbers)
```

```python
# Set operations
set_a = {1, 2, 3}
set_b = {3, 4, 5}
print("Union:", set_a | set_b)
print("Intersection:", set_a & set_b)
print("Difference (A - B):", set_a - set_b)
```

```python
#DICTIONARIES
student = {
    'name': 'Alice',
    'age': 21,
    'course': 'Python'
}
print("Dictionary:", student)
```

```python
# Access value by key
print("Student name:", student['name'])
```

```python
# Add or update value
student['grade'] = 'A'
print("After adding grade:", student)
# Remove a key
del student['age']
print("After deleting age:", student)
# Loop through dictionary
for key, value in student.items():
    print(f"{key}: {value}")
```

```python
# Bonus: Dictionary from two lists using zip()
keys = ['id', 'name', 'marks']
values = [101, 'John', 88]
student_info = dict(zip(keys, values))
print("Student Info from zip:", student_info)
```

```python
# Define a list
my_list = [10, 20, 30, 40, 50]
```

```python
# Indexing (0-based)
print("First element:", my_list[0])
print("Last element:", my_list[-1])
```

```python
# Slicing
print("Elements from index 1 to 3 (not inclusive):", my_list[1:4])
print("From start to index 2:", my_list[:3])
print("From index 2 to end:", my_list[2:])
print("Every second element:", my_list[::2])
print("Reversed list:", my_list[::-1])
```

```python
# Define a tuple
my_tuple = ("a", "b", "c", "d", "e")
```

```python
# Indexing
print("Element at index 0:", my_tuple[0])
print("Element at index -2:", my_tuple[-2])  # 'd'
```

```python
# Slicing
print("Slice from index 1 to 3:", my_tuple[1:4])
print("Tuple reversed:", my_tuple[::-1])
```

```python
# Define a string
text = "hello world"
```

```python
# Indexing
print("Character at index 0:", text[0])
print("Last character:", text[-1])
```

```python
# Slicing
print("First 6 characters:", text[:6])
print("From 6 to end:", text[6:])
print("Every 2nd character:", text[::2])
print("Reversed string:", text[::-1])
```
