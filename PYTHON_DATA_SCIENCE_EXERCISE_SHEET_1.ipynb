**Part C – Measures of Central Tendency**

```python
#A. Mean
print("Age Mean:", df["Age"].mean())
print("Experience Mean:", df["Experience"].mean())
print("Salary Mean:", df["Salary"].mean())
print("Performance Score Mean:", df["Performance_Score"].mean())
```

```python
#B. Median
print("Age Median:", df["Age"].median())
print("Experience Median:", df["Experience"].median())
print("Salary Median:", df["Salary"].median())
print("Performance Score Median:", df["Performance_Score"].median())
```

```python
#C. Mode
print("Age Mode:\n", df["Age"].mode().to_list())
print("Experience Mode:\n", df["Experience"].mode().to_list())
print("Salary Mode:\n", df["Salary"].mode().to_list())
print("Performance Score Mode:\n", df["Performance_Score"].mode().to_list())
```

**Question: Which variable has the largest mean?**

Salary has the largest mean with a value of ₹57,666.67, which is much higher than any other variable's mean.

**Question: Which variable has the smallest median?**

Experience has the smallest median with a value of 5.0, reflecting the lower middle boundary of years worked.

**Question: For which variable are the mean and median substantially different?**

Salary shows a substantial difference between its mean of ₹57,666.67 and its median of ₹40,000.00, along with Experience which has a mean of 7.53 and a median of 5.0.

**Question: Why might the mean salary be affected by very high salaries?**

The mean is calculated by summing every value and dividing by the total count, which causes extreme high values to pull the overall average upward away from the majority of the data points.

**Part D – Measures of Spread / Dispersion**

```python
#A. Range
age_range = df["Age"].max() - df["Age"].min()
print(age_range)
```

```python
#B. Variance
print("Age Variance:", df["Age"].var())
print("Experience Variance:", df["Experience"].var())
print("Salary Variance:", df["Salary"].var())
print("Performance Score Variance:", df["Performance_Score"].var())
```

```python
#C. Standard Deviation
print("Age Standard Deviation:", df["Age"].std())
print("Experience Standard Deviation:", df["Experience"].std())
print("Salary Standard Deviation:", df["Salary"].std())
print("Performance Score Standard Deviation:", df["Performance_Score"].std())
```

```python
#D. Quartiles And E. IQR (Example For Salary)
q1 = df["Salary"].quantile(0.25)
q3 = df["Salary"].quantile(0.75)
iqr = q3 - q1
print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
```

**Question: Which variable has the largest range?**

Salary has the largest range with a value of ₹125,000, representing the distance between the lowest salary of ₹25,000 and the highest salary of ₹150,000.

**Question: Which variable has the largest standard deviation?**

Salary has the largest standard deviation with a value of ₹34,472.90, showing the greatest spread relative to its mean.

**Question: What does a large standard deviation indicate?**

A large standard deviation indicates that the individual data points are widely scattered and spread far away from the mean score rather than being closely clustered around it.

**Question: What does IQR measure?**

The IQR measures the range and spread of the middle 50% of the dataset by subtracting the first quartile from the third quartile ($Q_3 - Q_1$).

**Question: Why is IQR less affected by extreme values than range?**
The IQR is less affected because it only focuses on the middle 50% of ordered observations, completely trimming off extreme upper and lower outliers before calculating spread.

**Part E – Skewness**

```python
#Calculate Skewness For Numerical Variables
skewness = df[["Age", "Experience", "Salary", "Performance_Score"]].skew()
print(skewness)
```

**Question: Which variable has positive skewness?**

Salary (1.62), Age (0.98), and Experience (0.94) all exhibit positive skewness values greater than zero.

**Question: Which variable has negative skewness?**

None of the numerical variables in this dataset exhibit negative skewness, as all skewness coefficients are greater than zero.

**Question: Which variable is approximately symmetric?**

Performance_Score is approximately symmetric because its skewness coefficient of 0.22 is very close to zero.

**Question: Does salary appear to be right-skewed?**

Yes, salary appears strongly right-skewed because its positive skewness value of 1.62 indicates a longer right tail toward high values.

**Question: What is the relationship between mean and median for a positively skewed variable?**

For a positively skewed variable, the mean is greater than the median because extreme upper values pull the average upward.

**Part G – Combined Statistical Summary**

```python
#Summary
summary = pd.DataFrame({
    "Mean": [
        df["Age"].mean(),
        df["Experience"].mean(),
        df["Salary"].mean(),
        df["Performance_Score"].mean()
    ],
    "Median": [
        df["Age"].median(),
        df["Experience"].median(),
        df["Salary"].median(),
        df["Performance_Score"].median()
    ],
    "Std_Deviation": [
        df["Age"].std(),
        df["Experience"].std(),
        df["Salary"].std(),
        df["Performance_Score"].std()
    ],
    "Skewness": [
        df["Age"].skew(),
        df["Experience"].skew(),
        df["Salary"].skew(),
        df["Performance_Score"].skew()
    ]
})
summary.index = ["Age", "Experience", "Salary", "Performance Score"]
print(summary.round(2))
```

**Part H – Data Science Interpretation**



**Question: The salary distribution contains a few very high salaries. Is it likely to be positively or negatively skewed?**

It is likely to be positively skewed because extreme high values pull the tail of the distribution to the right.



**Question: Which is likely to be greater for a positively skewed salary distribution: mean or median?**

The mean is likely to be greater than the median because high earners inflate the total sum used to calculate the mean.


**Question: Remove the employee earning ₹150,000 and calculate the new mean, median, standard deviation, and skewness. Compare with the original dataset.**

Removing the ₹150,000 salary drops the mean from ₹57,666.67 to ₹51,071.43, slightly decreases the median from ₹40,000.00 to ₹38,000.00, reduces the standard deviation from ₹34,472.90 to ₹22,359.39, and cuts skewness in half from 1.62 down to 0.81.


**Question: Which measure of central tendency would be more representative of salary when extreme salaries exist? Explain using your results.**

The median is more representative because it remains steady at ₹40,000.00 near the majority of employees, whereas the mean of ₹57,666.67 is artifically inflated by a single outlier even though 10 out of 15 employees make ₹50,000 or less.


**Question: Which measure of spread would you prefer when extreme values are present: range, standard deviation, or IQR? Explain.**

The IQR is preferred because it focuses purely on the middle 50% of data and remains stable, while range and standard deviation are heavily distorted by extreme values.

**Part J – Final Challenge**

```python
numeric_columns = df.select_dtypes(include=np.number).columns
numeric_columns = [col for col in numeric_columns if col != "Employee_ID"]
for column in numeric_columns:
    q1 = df[column].quantile(0.25)
    q3 = df[column].quantile(0.75)
    iqr = q3 - q1
    print(f"\n--- {column} ---")
    print("Mean:", round(df[column].mean(), 2))
    print("Median:", df[column].median())
    print("Mode:", df[column].mode().tolist())
    print("Minimum:", df[column].min())
    print("Maximum:", df[column].max())
    print("Range:", df[column].max() - df[column].min())
    print("Variance:", round(df[column].var(), 2))
    print("Standard Deviation:", round(df[column].std(), 2))
    print("Q1:", q1)
    print("Q3:", q3)
    print("IQR:", iqr)
    print("Skewness:", round(df[column].skew(), 2))
```
