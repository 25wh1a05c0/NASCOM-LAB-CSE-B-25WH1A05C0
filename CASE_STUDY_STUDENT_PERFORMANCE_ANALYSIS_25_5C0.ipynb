**Part G – Calculus: Integration**

**Task 7: Calculate ∫₀⁵ L(x) dx Where Learning Benefit Is Represented By: L(x) = 5x − 0.2x²**

```python
# 24. Solve The Integral Manually
import sympy as sp
x = sp.symbols('x')
L = 5*x - 0.2*x**2
integral = sp.integrate(L, (x, 0, 5))
print("Integral =", integral)
print("Approximate value =", float(integral))
```

```python
# 25. Implement A Numerical Approximation Using Python
import numpy as np
def L(x):
    return 5*x - 0.2*x**2
x = np.linspace(0, 5, 1000)
y = L(x)
area = np.trapezoid(y, x)
print("Numerical approximation =", area)
```

```python
# 26. Explain What The Area Under The Curve Represents In This Case
exact_area = 54.16666666666667
print("Area under the learning-benefit curve =", round(exact_area, 2))
print("This represents the total accumulated learning benefit")
print("from 0 to 5 hours of study.")
```
