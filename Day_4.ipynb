```python
#1
import numpy as np
import sympy as sp
def g(x):
    return x**3 + 2*x

# 1. Numerical derivative at x = 2 (use h = 1e-6)
# YOUR CODE HERE
h = 1e-6
x = 2
numerical_derivative = (g(x + h) - g(x - h)) / (2 * h)
print("Numerical derivative:", numerical_derivative)
# 2. Symbolic derivative of x**3 + 2*x
# YOUR CODE HERE
x_sym = sp.symbols('x')
g_sym = x_sym**3 + 2*x_sym
symbolic_derivative = sp.diff(g_sym, x_sym)
print("Symbolic derivative:", symbolic_derivative)
# 3. Evaluate the symbolic derivative at x = 2 and compare
# YOUR CODE HERE
symbolic_value = symbolic_derivative.subs(x_sym, 2)
print("Symbolic derivative at x = 2:", symbolic_value)
print("Difference:", abs(numerical_derivative - float(symbolic_value)))
```

```python
#2
import numpy as np
import sympy as sp

x, y = sp.symbols('x y')
h2 = x**2 * y + sp.sin(y)

# 1. dh/dx and dh/dy
# YOUR CODE HERE
dh_dx = sp.diff(h2, x)
dh_dy = sp.diff(h2, y)
print("dh/dx =", dh_dx)
print("dh/dy =", dh_dy)
# 2. Assemble the gradient list
# YOUR CODE HERE
gradient = [dh_dx, dh_dy]
print("Gradient =", gradient)
# 3. Evaluate at (x=2, y=0)  -> hint: .subs({x: 2, y: 0})
# YOUR CODE HERE
gradient_at_point = [expr.subs({x: 2, y: 0}) for expr in gradient]
print("Gradient at (2, 0) =", gradient_at_point)
```

```python
#4
import numpy as np
import sympy as sp
Xb = np.random.randn(6, 4)
Yb = np.random.randn(6, 1)
Wa = np.random.randn(4, 8) * 0.1     # input -> hidden
Wb = np.random.randn(8, 1) * 0.1     # hidden -> output

# 1. Forward pass: z1, h = ReLU(z1), y_hat, loss
# YOUR CODE HERE
z1 = Xb @ Wa
h = np.maximum(0, z1)
y_hat = h @ Wb
loss_before = np.mean((y_hat - Yb) ** 2)
print("Loss before:", loss_before)
# 2. Backward pass: dy, dWb, dh, dz1, dWa
# YOUR CODE HERE
dy = (2 / Yb.shape[0]) * (y_hat - Yb)
dWb = h.T @ dy
dh = dy @ Wb.T
dz1 = dh * (z1 > 0)
dWa = Xb.T @ dz1
# 3. One step with lr = 0.05; print loss before and after
# YOUR CODE HERE
lr = 0.05
Wa -= lr * dWa
Wb -= lr * dWb
z1 = Xb @ Wa
h = np.maximum(0, z1)
y_hat = h @ Wb
loss_after = np.mean((y_hat - Yb) ** 2)
print("Loss before:", loss_before)
print("Loss after:", loss_after)
```
